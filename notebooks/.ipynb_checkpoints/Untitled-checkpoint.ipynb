```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
u = 1/np.array([20, 32, 40, 50])
E = [-0.24856, -0.30130, -0.32079, -0.33810]
plt.plot(u, E)
plt.show()
```

```python
c1, c0 = np.polyfit(u, E, 1)
s = (E[-1] - E[-2]) / (u[-1] - u[-2]); b = E[-1] - s * u[-1]   # finest pair
print(f"4-point E_inf = {c0:.4f} eV | 40-50 pair E_inf = {b:.4f} eV")
```

```python
us = np.linspace(0, u.max() * 1.05, 100)
plt.plot(u, E, "ko", label="ironmc")
plt.plot(us, c0 + c1 * us, "-",  label=f"4-pt fit: {c0:.3f} eV")
plt.plot(us, b + s * us,  "--", label=f"40³–50³: {b:.3f} eV")
plt.axhline(-0.408, color="r", ls=":", label="FEP-DMC ∞ (−0.408)")
plt.plot(0.05, -0.2479, "r^", label="FEP-DMC 20³")

plt.xlabel("u = 1/n"); plt.ylabel(r"$E_0$ (eV)"); plt.xlim(left=0)
plt.legend(); plt.tight_layout(); plt.savefig("ladder.png", dpi=150); plt.show()
```

$$E_0(u) = E_\infty + C u + D u^2 + \dots$$

```python
print("quad  E_inf =", np.polyfit(u,E,2)[-1])          # -0.423 (over-fits 4 pts, 1 dof)
for k in range(3):                                      # progressive drop
    m=slice(k,None); print(f"drop{k}: linear E_inf =", np.polyfit(u[m],E[m],1)[1])
```
